```python
from transformers import AutoTokenizer, AutoModelForSequenceClassification

model_name = "distilbert-base-uncased"

tokenizer = AutoTokenizer.from_pretrained(model_name)

model = AutoModelForSequenceClassification.from_pretrained(
    model_name,
    num_labels=5
)

print("Model loaded successfully!")
```

```python
import torch

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")

model = model.to(device)

print("Using device:", device)
```

```python
from datasets import load_dataset

dataset = load_dataset("dipenbhuva/home-diy-repair-qa")

train = dataset["train"]
eval_data = dataset["eval"]

print("Training examples:", len(train))
print("Evaluation examples:", len(eval_data))
```

```python
from collections import Counter
from sklearn.model_selection import train_test_split

label_names = [
    "appliance_repair",
    "electrical_repair",
    "general_home_repair",
    "hvac_maintenance",
    "plumbing_repair"
]

label_to_id = {
    label: i
    for i, label in enumerate(label_names)
}

unique_data = {}

for row in train:
    question = row["question"]

    if question not in unique_data:
        unique_data[question] = row

clean_train = list(unique_data.values())

texts = []
labels = []

for row in clean_train:
    texts.append(row["question"])
    labels.append(label_to_id[row["category"]])

train_texts, temp_texts, train_labels, temp_labels = train_test_split(
    texts,
    labels,
    test_size=0.20,
    random_state=42,
    stratify=labels
)

val_texts, test_texts, val_labels, test_labels = train_test_split(
    temp_texts,
    temp_labels,
    test_size=0.50,
    random_state=42,
    stratify=temp_labels
)

print("Unique questions:", len(clean_train))
print("Training:", len(train_texts))
print("Validation:", len(val_texts))
print("Test:", len(test_texts))
```

```python
train_encodings = tokenizer(
    train_texts,
    truncation=True,
    padding=True,
    max_length=64
)

val_encodings = tokenizer(
    val_texts,
    truncation=True,
    padding=True,
    max_length=64
)

test_encodings = tokenizer(
    test_texts,
    truncation=True,
    padding=True,
    max_length=64
)

print("Training examples:", len(train_encodings["input_ids"]))
print("Validation examples:", len(val_encodings["input_ids"]))
print("Test examples:", len(test_encodings["input_ids"]))
```

```python
import torch

class HomeAssistDataset(torch.utils.data.Dataset):
    def __init__(self, encodings, labels):
        self.encodings = encodings
        self.labels = labels

    def __len__(self):
        return len(self.labels)

    def __getitem__(self, index):
        item = {
            key: torch.tensor(value[index])
            for key, value in self.encodings.items()
        }

        item["labels"] = torch.tensor(self.labels[index])

        return item


train_dataset = HomeAssistDataset(train_encodings, train_labels)
val_dataset = HomeAssistDataset(val_encodings, val_labels)
test_dataset = HomeAssistDataset(test_encodings, test_labels)

print("Training dataset:", len(train_dataset))
print("Validation dataset:", len(val_dataset))
print("Test dataset:", len(test_dataset))
```

```python
from transformers import TrainingArguments

training_args = TrainingArguments(
    output_dir="./homeassist_distilbert",
    eval_strategy="epoch",
    save_strategy="epoch",
    learning_rate=2e-5,
    per_device_train_batch_size=16,
    per_device_eval_batch_size=16,
    num_train_epochs=4,
    weight_decay=0.01,
    load_best_model_at_end=True,
    metric_for_best_model="f1",
    greater_is_better=True,
    report_to="none"
)

print("Training configuration created successfully!")
```

```python
import numpy as np
from sklearn.metrics import accuracy_score, precision_recall_fscore_support

def compute_metrics(eval_pred):
    logits, labels = eval_pred

    predictions = np.argmax(logits, axis=-1)

    precision, recall, f1, _ = precision_recall_fscore_support(
        labels,
        predictions,
        average="macro",
        zero_division=0
    )

    accuracy = accuracy_score(labels, predictions)

    return {
        "accuracy": accuracy,
        "precision": precision,
        "recall": recall,
        "f1": f1
    }

print("Evaluation metrics ready!")
```

```python
from transformers import Trainer

trainer = Trainer(
    model=model,
    args=training_args,
    train_dataset=train_dataset,
    eval_dataset=val_dataset,
    compute_metrics=compute_metrics
)

print("Trainer created successfully!")
```

```python
print("Device:", device)
print("Training samples:", len(train_dataset))
print("Validation samples:", len(val_dataset))
print("Test samples:", len(test_dataset))
print("Number of labels:", model.config.num_labels)
print("Labels:", label_names)
```

```python
trainer.train()
```

```python
test_results = trainer.evaluate(test_dataset)

print(test_results)
```
